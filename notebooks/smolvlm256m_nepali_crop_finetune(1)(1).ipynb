```python
import unsloth
from unsloth import FastVisionModel

import os, json, random, time, zipfile
from pathlib import Path

import torch
from PIL import Image
from huggingface_hub import HfApi, hf_hub_download
from datasets import Dataset as HFDataset
from torch.utils.data import Dataset
from transformers import (
    TrainingArguments,
    Trainer,
    TrainerCallback,
)

torch.manual_seed(42)
random.seed(42)

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
print("Using device:", DEVICE)
```

```python
MODEL_ID = "HuggingFaceTB/SmolVLM-256M-Instruct"
DATASET_REPO = "w4ashabii/nepali_crop_data"

DATA_DIR = Path("/content/nepali_crop_data")
OUTPUT_DIR = "/content/smolvlm-256m-nepali-crop-lora"

TRAIN_SUBSET_SIZE = None
EVAL_SUBSET_SIZE = 500

IMAGE_LONGEST_EDGE = 512
DO_IMAGE_SPLITTING = False

PER_DEVICE_BATCH_SIZE = 8
GRAD_ACCUM_STEPS = 2
NUM_EPOCHS = 2
LEARNING_RATE = 1e-4

SAVE_STEPS = 500

HUB_UPLOAD_REPO = "w4ashabii/SmolVLM256M_CropDisease"
UPLOAD_INTERVAL_SECONDS = 60 * 60
```

## 5. Download the dataset


```python
api = HfApi()
all_files = api.list_repo_files(DATASET_REPO, repo_type="dataset")
zip_files = sorted(f for f in all_files if f.startswith("zips/") and f.endswith(".zip"))
print(f"Found {len(zip_files)} zip part(s):")
for f in zip_files:
    print(" -", f)

DATA_DIR.mkdir(parents=True, exist_ok=True)
for zf_name in zip_files:
    local_zip = hf_hub_download(repo_id=DATASET_REPO, repo_type="dataset", filename=zf_name)
    print("Extracting", zf_name, "...")
    with zipfile.ZipFile(local_zip) as zf:
        zf.extractall(DATA_DIR)

print("Done. Contents of DATA_DIR:")
print(list(DATA_DIR.glob("*")))
```

```python
sharegpt_path = DATA_DIR / "sharegpt.jsonl"

records = []
with open(sharegpt_path, "r") as f:
    for line in f:
        line = line.strip()
        if line:
            records.append(json.loads(line))

print("Total conversations in sharegpt.jsonl:", len(records))
print("Example record:")
print(json.dumps(records[0], indent=2))
```

## 6. Build train/eval splits



```python
def resolve_image_path(rel_path: str) -> Path:
    return DATA_DIR / rel_path

def filter_and_resolve(records, split_prefix):
    out = []
    for r in records:
        img_rel = r["image"]
        if not img_rel.startswith(split_prefix):
            continue
        img_path = resolve_image_path(img_rel)
        if img_path.exists():
            out.append({"image_path": str(img_path), "conversations": r["conversations"]})
    return out

train_records = filter_and_resolve(records, "train/")
val_records = filter_and_resolve(records, "val/")

print("Resolved train examples:", len(train_records))
print("Resolved val examples:", len(val_records))

random.shuffle(train_records)
random.shuffle(val_records)

if TRAIN_SUBSET_SIZE is not None:
    train_records = train_records[:TRAIN_SUBSET_SIZE]
if EVAL_SUBSET_SIZE is not None:
    val_records = val_records[:EVAL_SUBSET_SIZE]

print("Using train examples:", len(train_records))
print("Using eval examples:", len(val_records))
```

## 7. `Dataset` + collate function



```python
class CropDiseaseDataset(Dataset):
    def __init__(self, records):
        self.records = records

    def __len__(self):
        return len(self.records)

    def __getitem__(self, idx):
        r = self.records[idx]
        image = Image.open(r["image_path"]).convert("RGB")
        return {"image": image, "conversations": r["conversations"]}


train_dataset = CropDiseaseDataset(train_records)
eval_dataset = CropDiseaseDataset(val_records)

len(train_dataset), len(eval_dataset)
```

```python
IMAGE_TOKEN_ID = None
```

```python
def build_collate_fn(processor, image_token_id):
    def collate_fn(examples):
        texts, images = [], []
        for ex in examples:
            human_turn, gpt_turn = ex["conversations"][0], ex["conversations"][1]
            question = human_turn["value"].replace("<image>", "").strip()
            answer = gpt_turn["value"].strip()

            messages = [
                {"role": "user", "content": [{"type": "image"}, {"type": "text", "text": question}]},
                {"role": "assistant", "content": [{"type": "text", "text": answer}]},
            ]
            text = processor.apply_chat_template(messages, add_generation_prompt=False).strip()
            texts.append(text)
            images.append([ex["image"]])

        batch = processor(text=texts, images=images, return_tensors="pt", padding=True)

        labels = batch["input_ids"].clone()
        labels[labels == processor.tokenizer.pad_token_id] = -100
        if image_token_id is not None:
            labels[labels == image_token_id] = -100
        batch["labels"] = labels
        return batch

    return collate_fn
```

## 8. Load the model and attach LoRA adapters (via Unsloth)



```python
model, processor = FastVisionModel.from_pretrained(
    MODEL_ID,
    load_in_4bit=False,                     # SmolVLM-256M is tiny -- full precision LoRA is already lightweight on a T4
    use_gradient_checkpointing="unsloth",   # Unsloth's optimized checkpointing: less VRAM, faster than "true"
)

processor.image_processor.size = {"longest_edge": IMAGE_LONGEST_EDGE}
processor.image_processor.do_image_splitting = DO_IMAGE_SPLITTING
processor.tokenizer.padding_side = "right"
if processor.tokenizer.pad_token_id is None:
    processor.tokenizer.pad_token = processor.tokenizer.eos_token

IMAGE_TOKEN_ID = getattr(model.config, "image_token_id", None)
print("image_token_id:", IMAGE_TOKEN_ID)

model = FastVisionModel.get_peft_model(
    model,
    finetune_vision_layers=True,
    finetune_language_layers=True,
    finetune_attention_modules=True,
    finetune_mlp_modules=True,
    r=8,
    lora_alpha=16,
    lora_dropout=0.05,
    bias="none",
    random_state=42,
)

FastVisionModel.for_training(model)  # put the model in train mode (grads on, dropout active, etc.)
model.print_trainable_parameters()
```

## 8b. Hourly Hugging Face Hub backups


```python
class HourlyHubUploadCallback(TrainerCallback):
    """Pushes `output_dir` to `repo_id` on the Hub every `interval_seconds`,
    deleting the repo's existing files first so it mirrors the latest checkpoint
    instead of accumulating old ones."""

    def __init__(self, output_dir, repo_id, interval_seconds=3600, repo_type="model"):
        self.output_dir = output_dir
        self.repo_id = repo_id
        self.interval_seconds = interval_seconds
        self.repo_type = repo_type
        self.api = HfApi()
        self.api.create_repo(repo_id, repo_type=repo_type, exist_ok=True)
        self.last_upload = time.time()

    def upload_now(self, label="checkpoint"):
        print(f"[{time.strftime('%H:%M:%S')}] Uploading {label} to {self.repo_id} "
              f"(deleting previous files first) ...")
        try:
            self.api.upload_folder(
                folder_path=self.output_dir,
                repo_id=self.repo_id,
                repo_type=self.repo_type,
                delete_patterns=["*"],      # remove everything currently in the repo, in the same commit
                commit_message=f"Update: {label}",
            )
            print("  upload complete.")
        except Exception as e:
            print("  upload failed:", e)
        self.last_upload = time.time()

    def on_step_end(self, args, state, control, **kwargs):
        if time.time() - self.last_upload >= self.interval_seconds:
            self.upload_now(label=f"step {state.global_step}")

    def on_train_end(self, args, state, control, **kwargs):
        self.upload_now(label=f"final checkpoint, step {state.global_step}")


hourly_uploader = HourlyHubUploadCallback(
    output_dir=OUTPUT_DIR,
    repo_id=HUB_UPLOAD_REPO,
    interval_seconds=UPLOAD_INTERVAL_SECONDS,
)
```

## 9. Benchmark: estimate total training time



```python
collate_fn = build_collate_fn(processor, IMAGE_TOKEN_ID)

_bench_args = TrainingArguments(
    output_dir="/content/_bench",
    per_device_train_batch_size=PER_DEVICE_BATCH_SIZE,
    gradient_accumulation_steps=GRAD_ACCUM_STEPS,
    max_steps=20,
    logging_steps=100,  # quiet
    fp16=True,
    dataloader_num_workers=2,
    remove_unused_columns=False,
    report_to="none",
    save_strategy="no",
)

_bench_trainer = Trainer(
    model=model,
    args=_bench_args,
    train_dataset=train_dataset,
    data_collator=collate_fn,
)

start = time.time()
_bench_trainer.train()
elapsed = time.time() - start
sec_per_optim_step = elapsed / 20  # 20 optimizer steps

n_train = len(train_dataset)
steps_per_epoch = max(1, n_train // (PER_DEVICE_BATCH_SIZE * GRAD_ACCUM_STEPS))
total_steps = steps_per_epoch * NUM_EPOCHS
est_total_seconds = total_steps * sec_per_optim_step

print(f"~{sec_per_optim_step:.2f} sec/optimizer-step at batch={PER_DEVICE_BATCH_SIZE}x{GRAD_ACCUM_STEPS}")
print(f"Train examples: {n_train} -> {steps_per_epoch} steps/epoch x {NUM_EPOCHS} epochs = {total_steps} steps")
print(f"Estimated full training time: {est_total_seconds/3600:.2f} hours")
print("(Add ~10-20% for eval passes + checkpoint saving.)")
print()
print("If this is too long for your session: lower TRAIN_SUBSET_SIZE and re-run from the")
print("'Build train/eval splits' cell, or lower NUM_EPOCHS, then re-create train_dataset/eval_dataset above.")
```

## 10. Train



```python
training_args = TrainingArguments(
    output_dir=OUTPUT_DIR,
    per_device_train_batch_size=PER_DEVICE_BATCH_SIZE,
    per_device_eval_batch_size=PER_DEVICE_BATCH_SIZE,
    gradient_accumulation_steps=GRAD_ACCUM_STEPS,
    num_train_epochs=NUM_EPOCHS,
    learning_rate=LEARNING_RATE,
    warmup_ratio=0.03,
    lr_scheduler_type="cosine",
    logging_steps=25,
    eval_strategy="steps",
    eval_steps=SAVE_STEPS,
    save_strategy="steps",
    save_steps=SAVE_STEPS,
    save_total_limit=3,
    fp16=True,
    dataloader_num_workers=2,
    remove_unused_columns=False,
    report_to="none",
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=eval_dataset,
    data_collator=collate_fn,
    callbacks=[hourly_uploader],   # hourly Hub backups, see Section 8b
)

trainer.train()  # use trainer.train(resume_from_checkpoint=True) to resume after a disconnect
```

## 11. Save the adapter

```python
from huggingface_hub import HfApi, CommitOperationDelete

api = HfApi()


repo_files = api.list_repo_files(HUB_UPLOAD_REPO, repo_type="model")
stale_files = [f for f in repo_files if f.startswith("checkpoint-")]

print(f"Found {len(stale_files)} stale checkpoint file(s):")
for f in stale_files:
    print(" -", f)

if stale_files:
    api.create_commit(
        repo_id=HUB_UPLOAD_REPO,
        repo_type="model",
        operations=[CommitOperationDelete(path_in_repo=f) for f in stale_files],
        commit_message="Remove stale checkpoint-* files",
    )
    print("Stale checkpoints deleted.")
else:
    print("No stale checkpoint files found.")


trainer.save_model(OUTPUT_DIR)
processor.save_pretrained(OUTPUT_DIR)

api.upload_folder(
    folder_path=OUTPUT_DIR,
    repo_id=HUB_UPLOAD_REPO,
    repo_type="model",
    delete_patterns=["*"],
    ignore_patterns=["checkpoint-*"],
    commit_message="Final adapter + processor",
)

print(f"Done: https://huggingface.co/{HUB_UPLOAD_REPO}")
```

## 12. Try it out



```python
FastVisionModel.for_inference(model)

def predict(image_path, question="What crop is shown in this image, and does it have any disease? If so, name the disease."):
    image = Image.open(image_path).convert("RGB")
    messages = [
        {"role": "user", "content": [{"type": "image"}, {"type": "text", "text": question}]}
    ]
    prompt = processor.apply_chat_template(messages, add_generation_prompt=True)
    inputs = processor(text=prompt, images=[image], return_tensors="pt").to(DEVICE)

    with torch.no_grad():
        generated_ids = model.generate(**inputs, max_new_tokens=64)

    output = processor.batch_decode(
        generated_ids[:, inputs["input_ids"].shape[1]:], skip_special_tokens=True
    )[0]
    return image, output.strip()


for r in val_records[:5]:
    img, answer = predict(r["image_path"])
    ground_truth = r["conversations"][1]["value"]
    print("Image:", r["image_path"])
    print("  Model prediction:", answer)
    print("  Ground truth    :", ground_truth)
    print()
```
